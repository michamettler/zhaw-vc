```python
from deepface import DeepFace
import cv2
import matplotlib.pyplot as plt
```

```python
def verify(img1_path, img2_path):
    img1 = cv2.imread(img1_path)
    img2 = cv2.imread(img2_path)

    plt.imshow(img1[:,:,::-1])
    plt.show()
    plt.imshow(img2[:,:,::-1])
    plt.show()

    result = DeepFace.verify(img1_path, img2_path)

    result
```

```python
#verify ("ressource/davis.jpg", "ressource/davis2.jpg")
DeepFace.verify(img1_path="ressource/davis.jpg", img2_path="ressource/davis2.jpg")
```
